YZ, çalışmayı desteklemekten ziyade ikame ettiğinde tükenmişlik tetiklenmektedir.
<br>2. <strong>YZ Kullanımı ve Beceri Kalıcılığı (Öğrenme Kaybı):</strong> <code>Weekly_GenAI_Hours</code> ve <code>Skill_Retention_Score</code> arasındaki negatif ilişki (-0.12) öğrenme süreçlerine dair alarm vermektedir. <br>3. <strong>Kullanım ve Bağımlılık Algısı:</strong> Haftalık YZ saati arttıkça bağımlılık algısının (1-10) doğrusal olarak arttığı (korelasyon: 0.67) ve bağımlılık algısı 7'yi geçen öğrencilerin ezici bir çoğunluğunun High riskli olduğu net bir şekilde doğrulanmıştır.

````python
# 2.9 Aykırı Değer Analizi (Tukey IQR Yöntemi)
outlier_summary = []
num_cols = [
    'Pre_Semester_GPA', 'Weekly_GenAI_Hours', 'Traditional_Study_Hours', 
    'Skill_Retention_Score', 'Anxiety_Level_During_Exams', 
    'Tool_Diversity', 'Perceived_AI_Dependency', 'Post_Semester_GPA'
]

for col in num_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    outlier_summary.append({
        'Değişken': col,
        'Alt Sınır': round(lower_bound, 3),
        'Üst Sınır': round(upper_bound, 3),
        'Aykırı Değer Sayısı': len(outliers),
        'Aykırı Değer Oranı (%)': round(len(outliers) / len(df) * 100, 3)
    })

outlier_df = pd.DataFrame(outlier_summary)
display(outlier_df.style.set_properties(**{'background-color': '#111827', 'color': '#f3f4f6', 'border-color': '#374151'})
    .set_table_styles([{'selector': 'th', 'props': [('background-color', '#1f2937'), ('color', '#a78bfa'), ('font-weight', 'bold')]}]))
print('Aykırı değer tablosu oluşturuldu.')
````

<p style="color:#fecdd3;font-size:0.83em;line-height:1.5;margin:0;">
Veri sızıntısı (data leakage), eğitim veri setinde modelin öğrenmemesi gereken, geleceğe dair veya tahmin anında (dönem başında) elde edilemeyecek bilgilerin yer alması durumudur.
````python
# ════════════════════════════════════════════════════════════════════
# 3.0.1 — GÜVENLİ TEMİZLİK (STATELESS) + CLEANED CSV
# Sadece dağılım öğrenmeyen dönüşümler. Imputation/Encoding/Scaling YOK.
# ════════════════════════════════════════════════════════════════════
from pathlib import Path

CLEANED_DIR  = Path('../data/processed')
CLEANED_DIR.mkdir(parents=True, exist_ok=True)
CLEANED_PATH = CLEANED_DIR / 'ai_student_impact_cleaned.csv'

df_safe = df.copy()
_n0 = len(df_safe)

# 1) Duplicate kaldırma (stateless)
df_safe = df_safe.drop_duplicates().reset_index(drop=True)
_n1 = len(df_safe)

# 2) Güvenli sütun çıkarma
SAFE_DROP_COLS = ['Öğrenci No', 'Dönem Öncesi GNO', 'Dönem Sonrası GNO']
df_safe = df_safe.drop(columns=[c for c in SAFE_DROP_COLS if c in df_safe.columns])

# 3) Dtype düzeltmesi: Ücretli Abonelik True/False → int 0/1
if 'Ücretli Abonelik' in df_safe.columns:
    df_safe['Ücretli Abonelik'] = df_safe['Ücretli Abonelik'].astype('boolean').astype('Int64')

# 4) İmkânsız değer temizliği — negatif saat = veri girişi hatası
for _col in ['Haftalık AI Saati', 'Geleneksel Çalışma Saati']:
    if _col in df_safe.columns:
        _mask = df_safe[_col] < 0
        if _mask.any():
            df_safe.loc[_mask, _col] = pd.NA

# 5) Kayıt — eksik değerler KORUNUR (pipeline impute edecek)
df_safe.to_csv(CLEANED_PATH, index=False, encoding='utf-8-sig')

print(f"✓ Cleaned CSV → {CLEANED_PATH}")
print(f"  Rows: {_n0} → {_n1} (after dedup), columns: {df_safe.shape[1]}")
print(f"  Missing cells preserved: {int(df_safe.isna().sum().sum()):,}")
display(df_safe.head(3))
````

````python
# ════════════════════════════════════════════════════════════════════
# 3 — Sütuna özel akıllı doldurma + NEDEN o yöntem?
# ════════════════════════════════════════════════════════════════════
strategy_rows = [{'Sütun': c,
                  'Eksik': int(df_clean[c].isna().sum()),
                  'Yöntem': STRATEGY_LABEL[c][0],
                  'Neden Bu Yöntem?': STRATEGY_LABEL[c][2]}
                 for c in IMPUTE_NUMERIC + IMPUTE_CATEG]
strategy_df = pd.DataFrame(strategy_rows)

def color_yontem(v):
    pal = {'Lineer Regresyon':'#22d3ee','KNN':'#a78bfa','Medyan':'#fbbf24'}
    return f'color: {pal.get(v,"#e2e8f0")}; font-weight: 600'

display(tbl_header('HANGİ SÜTUNU HANGİ YÖNTEMLE VE NEDEN DOLDURUYORUZ?', '#a78bfa'))
display(strategy_df.style
    .set_properties(**{'background-color':'#0f172a','color':'#e2e8f0',
                       'border':'1px solid #1e293b','padding':'10px 16px'})
    .set_properties(subset=['Neden Bu Yöntem?'], **{'text-align':'left'})
    .set_table_styles([{'selector':'th','props':[('background-color','#1e293b'),
                                                  ('color','#94a3b8'),
                                                  ('font-weight','700'),
                                                  ('padding','10px 16px'),
                                                  ('letter-spacing','0.5px'),
                                                  ('text-align','left')]}])
    .applymap(color_yontem, subset=['Yöntem'])
    .hide(axis='index'))

# Uygulama sırası: Median → KNN → Iterative → KNN Classifier
df_filled = df_clean.copy()
df_filled['Araç Çeşitliliği'] = df_filled['Araç Çeşitliliği'].fillna(df_filled['Araç Çeşitliliği'].median())
knn = KNNImputer(n_neighbors=5)
knn_out = pd.DataFrame(knn.fit_transform(df_filled[IMPUTE_NUMERIC]),
                       columns=IMPUTE_NUMERIC, index=df_filled.index)
for col in KNN_COLS: df_filled[col] = knn_out[col]
ii = IterativeImputer(random_state=42, max_iter=10, estimator=BayesianRidge())
ii_out = pd.DataFrame(ii.fit_transform(df_filled[IMPUTE_NUMERIC]),
                      columns=IMPUTE_NUMERIC, index=df_filled.index)
for col in LR_COLS: df_filled[col] = ii_out[col]

mask_known = df_filled['Prompt Yazma Becerisi'].notna()
scaler = StandardScaler()
X_known = scaler.fit_transform(df_filled.loc[mask_known, IMPUTE_NUMERIC])
y_known = df_filled.loc[mask_known, 'Prompt Yazma Becerisi']
knn_clf = KNeighborsClassifier(n_neighbors=5, weights='distance')
knn_clf.fit(X_known, y_known)
X_miss = scaler.transform(df_filled.loc[~mask_known, IMPUTE_NUMERIC])
df_filled.loc[~mask_known, 'Prompt Yazma Becerisi'] = knn_clf.predict(X_miss)

assert df_filled[ALL_MISS].isnull().sum().sum() == 0

orig_d  = df_clean['Prompt Yazma Becerisi'].dropna().value_counts(normalize=True)
final_d = df_filled['Prompt Yazma Becerisi'].value_counts(normalize=True)
max_cat_shift = float((final_d - orig_d).abs().max() * 100)


# ════════════════════════════════════════════════════════════════════
# 4 — Doldurulan değer NEDEN o değer oldu?  (yönteme özel kanıt grafiği)
#     Histogram yerine her yöntemin mantığını gösteren temsili grafik.
# ════════════════════════════════════════════════════════════════════
_rs = np.random.RandomState(42)

# ── 4a) LİNEER REGRESYON — değer ilişki çizgisinden "okunur"
# Tek örnek öğrenci üzerinden anlatım: bilinen sütundan çizgiye çıkıp
# eksik sütunun karşılığını okuyoruz. (Gerçek doldurma tüm sayısal
# sütunları kullanan IterativeImputer'dır; çizgi referans mantığını gösterir.)
xp_col, yp_col = 'Algılanan AI Bağımlılığı', 'Haftalık AI Saati'
both = df_clean[[xp_col, yp_col]].notna().all(axis=1).values
xv = df_clean.loc[both, xp_col].values
yv = df_clean.loc[both, yp_col].values
slope, intercept = np.polyfit(xv, yv, 1)
r_lr = float(np.corrcoef(xv, yv)[0, 1])

n_samp = min(2500, int(both.sum()))
s_idx = _rs.choice(np.where(both)[0], size=n_samp, replace=False)
line_x = np.array([xv.min(), xv.max()])
line_y = slope * line_x + intercept

# Örnek eksik öğrenci: yp eksik, xp gözlenmiş; xp'si 75. persentile yakın olan
ex_pool = np.where((df_clean[yp_col].isna() & df_clean[xp_col].notna()).values)[0]
target_x = np.percentile(xv, 75)
ex_i = int(ex_pool[np.argmin(np.abs(df_clean.iloc[ex_pool][xp_col].values - target_x))])
x0 = float(df_clean.iloc[ex_i][xp_col])
y0 = float(slope * x0 + intercept)
y_lo = float(min(yv.min(), 0))

fig_lr = go.Figure()
fig_lr.add_trace(go.Scatter(
    x=df_clean.iloc[s_idx][xp_col], y=df_clean.iloc[s_idx][yp_col],
    mode='markers', name='Gözlenen öğrenci',
    marker=dict(color='rgba(148,163,184,0.30)', size=5),
    hovertemplate='Gözlenen<br>bağımlılık %{x:.0f} · %{y:.1f} saat<extra></extra>'))
fig_lr.add_trace(go.Scatter(
    x=line_x, y=line_y, mode='lines', name=f'Regresyon çizgisi (r={r_lr:.2f})',
    line=dict(color='#22d3ee', width=3)))
# Okuma kılavuzu: x0'dan çizgiye çık, çizgiden y eksenine git
fig_lr.add_trace(go.Scatter(
    x=[x0, x0, line_x[0]], y=[y_lo, y0, y0], mode='lines',
    line=dict(color='#f97316', width=1.6, dash='dash'),
    name='Okuma kılavuzu', hoverinfo='skip'))
fig_lr.add_trace(go.Scatter(
    x=[x0], y=[y0], mode='markers+text', name='Eksik öğrenci → okunan değer',
    marker=dict(color='#f97316', size=16, symbol='star',
                line=dict(color='#fed7aa', width=1.5)),
    text=[f"  ≈ {y0:.1f} saat"], textposition='middle right',
    textfont=dict(color='#fdba74', size=13),
    hovertemplate=f'Eksik öğrenci · bağımlılık={x0:.0f}<br>çizgiden okunan ≈ {y0:.1f} saat<extra></extra>'))
fig_lr.update_layout(
    title=dict(text='<b>Lineer Regresyon · değer ilişki çizgisinden okunur</b>  <span style="color:#64748b; font-size:12px">· '+xp_col+' → '+yp_col+'</span>',
               font=dict(size=16, color=P['text'], family='Inter'), x=0.02, xanchor='left'),
    paper_bgcolor=P['bg'], plot_bgcolor=P['bg'],
    font=dict(family='Inter', color=P['text'], size=11),
    height=440, margin=dict(l=70, r=40, t=80, b=70),
    xaxis=dict(title=xp_col, gridcolor=P['grid'], zeroline=False),
    yaxis=dict(title=yp_col, gridcolor=P['grid'], zeroline=False),
    legend=dict(bgcolor='rgba(15,23,42,0.7)', bordercolor=P['grid'], borderwidth=1,
                orientation='h', y=-0.20, x=0.5, xanchor='center'),
)
fig_lr.show()
display(card('NEDEN BU DEĞER', '#22d3ee',
    f"İki sütun arasında güçlü doğrusal ilişki var (r={r_lr:.2f}); gözlenen öğrenciler bu çizgiyi "
    f"çiziyor. AI bağımlılığı {x0:.0f} olan eksik bir öğrencinin Haftalık AI Saati değeri, kesikli "
    f"kılavuzun gösterdiği gibi çizgiden <b>≈{y0:.1f}</b> olarak okunur."))

# ── 4b) KNN — örnek öğrenci + en yakın 5 komşu (2B PCA projeksiyonu)
knn_demo_col = 'Sınav Kaygı Düzeyi'
feat = [c for c in IMPUTE_NUMERIC if c != knn_demo_col]
donor_mask = (df_clean[knn_demo_col].notna() & df_clean[feat].notna().all(axis=1)).values
ex_pool = np.where((df_clean[knn_demo_col].isna() & df_clean[feat].notna().all(axis=1)).values)[0]
ex_i = int(ex_pool[0])

Xd = df_clean.loc[donor_mask, feat].values
sc_knn = StandardScaler().fit(Xd)
Xd_s = sc_knn.transform(Xd)
xe_s = sc_knn.transform(df_clean.iloc[[ex_i]][feat].values)

nn = NearestNeighbors(n_neighbors=5).fit(Xd_s)
_, nn_ind = nn.kneighbors(xe_s)
donor_pos = np.where(donor_mask)[0]
neigh_global = donor_pos[nn_ind[0]]
neigh_vals = df_clean.iloc[neigh_global][knn_demo_col].values
imp_val_knn = float(neigh_vals.mean())

pca = PCA(n_components=2).fit(Xd_s)
bg_n = min(1500, len(Xd_s))
bg_sel = _rs.choice(len(Xd_s), size=bg_n, replace=False)
bg2 = pca.transform(Xd_s[bg_sel])
pe2 = pca.transform(xe_s)[0]
pn2 = pca.transform(Xd_s[nn_ind[0]])

# Komşuluğa yakınlaş — yıldız + 5 komşu net görünsün
_ax = np.concatenate([pn2[:, 0], [pe2[0]]])
_ay = np.concatenate([pn2[:, 1], [pe2[1]]])
_padx = max((_ax.max() - _ax.min()) * 1.6, 0.6)
_pady = max((_ay.max() - _ay.min()) * 1.6, 0.6)
_knn_xr = [_ax.mean() - _padx, _ax.mean() + _padx]
_knn_yr = [_ay.mean() - _pady, _ay.mean() + _pady]

fig_knn = go.Figure()
fig_knn.add_trace(go.Scatter(
    x=bg2[:, 0], y=bg2[:, 1], mode='markers', name='Diğer öğrenciler',
    marker=dict(color='rgba(148,163,184,0.18)', size=4), hoverinfo='skip'))
for k in range(len(pn2)):
    fig_knn.add_trace(go.Scatter(
        x=[pe2[0], pn2[k, 0]], y=[pe2[1], pn2[k, 1]], mode='lines',
        line=dict(color='rgba(167,139,250,0.45)', width=1.5),
        showlegend=False, hoverinfo='skip'))
fig_knn.add_trace(go.Scatter(
    x=pn2[:, 0], y=pn2[:, 1], mode='markers+text', name='5 en yakın komşu',
    marker=dict(color='#a78bfa', size=14, line=dict(color='#ede9fe', width=1)),
    text=[f"{v:.0f}" for v in neigh_vals], textposition='top center',
    textfont=dict(color='#ede9fe', size=11),
    hovertemplate='Komşu · kaygı=%{text}<extra></extra>'))
fig_knn.add_trace(go.Scatter(
    x=[pe2[0]], y=[pe2[1]], mode='markers+text', name='Eksik öğrenci',
    marker=dict(color='#f97316', size=20, symbol='star',
                line=dict(color='#fed7aa', width=1.5)),
    text=[f" → {imp_val_knn:.1f}"], textposition='middle right',
    textfont=dict(color='#fdba74', size=12),
    hovertemplate=f'Eksik öğrenci<br>doldurulan kaygı={imp_val_knn:.1f}<extra></extra>'))
fig_knn.update_layout(
    title=dict(text='<b>KNN · doldurulan değer = en yakın 5 komşunun ortalaması</b>  <span style="color:#64748b; font-size:12px">· '+knn_demo_col+' · 2B PCA görünümü</span>',
               font=dict(size=16, color=P['text'], family='Inter'), x=0.02, xanchor='left'),
    paper_bgcolor=P['bg'], plot_bgcolor=P['bg'],
    font=dict(family='Inter', color=P['text'], size=11),
    height=440, margin=dict(l=60, r=40, t=80, b=60),
    xaxis=dict(title='PCA bileşen 1', gridcolor=P['grid'], zeroline=False, range=_knn_xr),
    yaxis=dict(title='PCA bileşen 2', gridcolor=P['grid'], zeroline=False, range=_knn_yr),
    legend=dict(bgcolor='rgba(15,23,42,0.7)', bordercolor=P['grid'], borderwidth=1,
                orientation='h', y=-0.18, x=0.5, xanchor='center'),
)
fig_knn.show()
display(card('NEDEN BU DEĞER', '#a78bfa',
    f"Eksik öğrenci (yıldız), tüm diğer sütunları en çok benzeyen 5 komşusuyla mor çizgilerle "
    f"bağlandı. Komşuların {knn_demo_col} değerleri "
    f"{', '.join(f'{v:.0f}' for v in neigh_vals)}; bunların ortalaması <b>{imp_val_knn:.1f}</b> "
    f"olarak dolduruldu. Tek bir sütuna değil, 'en benzer öğrenciler' bilgisine dayanır."))

# Sonraki adıma taşınacak çıktı için kayma istatistiği (kart gösterilmiyor, sadece raporda)
shifts = []
for col in IMPUTE_NUMERIC:
    nan_mask = df_clean[col].isna()
    orig_vals = df_clean.loc[~nan_mask, col]
    shifts.append(abs(df_filled[col].mean() - orig_vals.mean()) / orig_vals.std() * 100)
max_shift = max(shifts)


# Sonraki adımlara taşınacak çıktı
IMPUTATION_DECISION = {
    'drop_cols'      : DROP_COLS,
    'dropped_rows_3+': n_dropped_3,
    'strategy_groups': {
        'iterative_lr'   : LR_COLS,
        'knn_5'          : KNN_COLS,
        'median'         : MED_COLS,
        'knn_classifier' : IMPUTE_CATEG,
    },
    'kept_shape'     : df_filled.shape,
    'max_shift_pct'  : round(max_shift, 2),
    'cat_dist_shift' : round(max_cat_shift, 2),
}
print(f">> Doldurma tamam | shape={df_filled.shape} | sayısal max kayma=%{max_shift:.2f} std | "
      f"kategorik max kayma=%{max_cat_shift:.1f}")
````

````python
# ════════════════════════════════════════════════════════════════════
# 3.2 — AYKIRI DEĞER STRATEJİSİ
# İki yöntem: IQR (çeyrekler arası çit) ve Z-skor (±3 sigma).
# Her yöntemi AYRI grafikte, aykırıları kırmızıyla işaretleyerek
# inceliyoruz. Aykırısı olmayan sütun için grafik üretmiyoruz.
# ════════════════════════════════════════════════════════════════════

# ── 1) IQR ve Z ile aykırı ölçümü (yalnız bu 2 yöntem)
out_rows = []
iqr_fences = {}
for col in IMPUTE_NUMERIC:
    s = df_filled[col]
    q1, q3 = s.quantile(0.25), s.quantile(0.75)
    iqr = q3 - q1
    lo, hi = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    iqr_fences[col] = (lo, hi)
    m_iqr = (s < lo) | (s > hi)
    z = (s - s.mean()) / s.std()
    m_z = z.abs() > 3
    out_rows.append({
        'Sütun'        : col,
        'IQR aykırı'   : int(m_iqr.sum()),
        'IQR (%)'      : round(m_iqr.mean() * 100, 2),
        'Z>3 aykırı'   : int(m_z.sum()),
        'Z>3 (%)'      : round(m_z.mean() * 100, 2),
        'Durum'        : 'Temiz' if (m_iqr.sum() == 0 and m_z.sum() == 0) else 'Aykırı var',
    })
out_tbl = pd.DataFrame(out_rows)

# Aykırısı olan sütunlar (grafik yalnız bunlar için çizilir)
iqr_cols = [r['Sütun'] for r in out_rows if r['IQR aykırı'] > 0]
z_cols   = [r['Sütun'] for r in out_rows if r['Z>3 aykırı'] > 0]

C_NORM = 'rgba(100,116,139,0.35)'   # normal (çit/band içi)
C_OUT  = '#ef4444'                   # aykırı
C_BAND = 'rgba(52,211,153,0.16)'     # kabul bölgesi
_rs = np.random.RandomState(7)
cnt_map = {r['Sütun']: r for r in out_rows}

def _prop_sample(inside, outside, n_show=2000):
    # Oranı koruyarak örnekle: aykırılar gerçek nadirliğiyle görünsün
    total = len(inside) + len(outside)
    frac = min(1.0, n_show / max(total, 1))
    in_n  = min(len(inside),  max(1, int(round(len(inside) * frac))))
    out_n = min(len(outside), max(2, int(round(len(outside) * frac)))) if len(outside) else 0
    si = inside.sample(in_n, random_state=7)
    so = outside.sample(out_n, random_state=7) if out_n else outside
    return si, so

# ── 2) Teknik ölçüm tablosu (yalnız IQR ve Z)
def _color_pct(v):
    try: x = float(v)
    except: return ''
    if x == 0  : return 'color: #475569'
    if x < 1   : return 'color: #94a3b8'
    if x < 3   : return 'color: #fbbf24; font-weight: 600'
    return       'color: #ef4444; font-weight: 700'

# Tabloda yalnız aykırısı olan sütunlar; "Durum" kolonu gereksiz, çıkarıldı
disp_tbl = out_tbl[out_tbl['Durum'] != 'Temiz'].drop(columns=['Durum']).reset_index(drop=True)

display(section_header('AYKIRI DEĞER TESPİTİ ve ANALİZİ', '#22d3ee',
    'IQR ve Z-skor yöntemleriyle uç değerleri tespit ediyor, gerçek davranış mı yoksa veri hatası mı olduğuna karar veriyoruz.'))

display(tbl_header('TEKNİK ÖLÇÜM · IQR ve Z-skor', '#22d3ee'))
display(disp_tbl.style
    .set_properties(**{'background-color':'#0f172a','color':'#e2e8f0',
                       'border':'1px solid #1e293b','padding':'10px 16px'})
    .set_table_styles([{'selector':'th','props':[('background-color','#1e293b'),
                                                  ('color','#94a3b8'),('font-weight','700'),
                                                  ('padding','10px 16px'),('letter-spacing','0.5px')]}])
    .applymap(_color_pct, subset=['IQR (%)','Z>3 (%)'])
    .hide(axis='index'))

n_clean = int((out_tbl['Durum'] == 'Temiz').sum())
worst_col = out_tbl.loc[out_tbl['IQR (%)'].idxmax(), 'Sütun']
worst_iqr = float(out_tbl['IQR (%)'].max())
display(card('IQR ve Z BİZE NE SÖYLER', '#22d3ee',
    "<b>IQR</b> ortadaki %50'lik kütlenin sınırlarını çizip dışına taşanları işaretler. "
    "<b>Z-skoru</b> ise her değerin ortalamadan kaç adım uzaklaştığını söyler. "
    "<em>Örnek:</em> sınıfta herkes haftada 5–15 saat çalışırken biri 40 saat çalışıyorsa, "
    "IQR de Z-skoru da 'bu öğrenci uçta' der — ama bu bir hata değil, gerçek bir davranış olabilir."))
````

````python
# ════════════════════════════════════════════════════════════════════
# 3.3 — ENCODING
# OHE şişmesinden kaçınarak akılcı kategorik dönüşüm stratejisi:
#   - Sıralı/hiyerarşik → Ordinal (1 sütun, sıra korunur)
#   - Sırasız              → WoE Encoding (log-odds, 1 sütun)
#   - Boolean              → Tamsayı (0/1)
# ════════════════════════════════════════════════════════════════════
from sklearn.preprocessing import OrdinalEncoder
from category_encoders import WOEEncoder
from plotly.subplots import make_subplots
import numpy as np

ENC_STRATEGY = {
    'Okunan Bölüm'            : ('WoE Encoding', '#f472b6',
                                   'Sırasız 5 kategori — log-odds oranıyla kodla'),
    'Birincil Kullanım Amacı' : ('WoE Encoding', '#f472b6',
                                   'Sırasız 5 kategori — log-odds oranıyla kodla'),
    'Kurum Politikası'        : ('Ordinal',       '#a78bfa',
                                   'Yasak < İzinli < Teşvik (hiyerarşik)'),
    'Sınıf Düzeyi'            : ('Ordinal',       '#a78bfa',
                                   '1. Sınıf → Lisansüstü sıralı'),
    'Prompt Yazma Becerisi'   : ('Ordinal',       '#a78bfa',
                                   'Başlangıç → İleri sıralı'),
    'Ücretli Abonelik'        : ('Tamsayı',       '#fbbf24',
                                   'Doğru→1, Yanlış→0'),
}

ORDINAL_COLS   = ['Sınıf Düzeyi', 'Prompt Yazma Becerisi', 'Kurum Politikası']
ORDINAL_ORDERS = [
    ['1. Sınıf', '2. Sınıf', '3. Sınıf', '4. Sınıf', 'Lisansüstü'],
    ['Başlangıç', 'Orta', 'İleri'],
    ['Kesin Yasak', 'Atıfla İzinli', 'Aktif Teşvik'],
]
WOE_ENC_COLS = ['Okunan Bölüm', 'Birincil Kullanım Amacı']
BOOL_COLS    = ['Ücretli Abonelik']

display(section_header('ENCODING ve SCALING İŞLEMLERİ', '#22d3ee',
    'Kategorik sütunları sayısala çeviriyor, sonra tüm sütunları ortak ölçeğe getiriyoruz.'))

# ── 1) Encoding Stratejisi — Treemap (statik tablo yerine)
_tree_data = []
for col, (method, color, reason) in ENC_STRATEGY.items():
    _tree_data.append({'Yöntem': method, 'Sütun': col, 'Neden': reason, 'Renk': color, 'Adet': 1})
_tree_df = pd.DataFrame(_tree_data)

_METHOD_COLORS = {
    'WoE Encoding': '#f472b6',
    'Ordinal':      '#a78bfa',
    'Tamsayı':      '#fbbf24',
}

fig_TREE = px.treemap(
    _tree_df,
    path=['Yöntem', 'Sütun'],
    values='Adet',
    color='Yöntem',
    color_discrete_map=_METHOD_COLORS,
    custom_data=['Neden'],
    title='<b>Encoding Stratejisi</b> — Hangi sütun, hangi yöntemle?',
)
fig_TREE.update_traces(
    hovertemplate='<b>%{label}</b><br><span style="color:#94a3b8">%{customdata[0]}</span><extra></extra>',
    textfont=dict(family='Inter', size=12, color='#0f172a'),
    marker_line=dict(color='#0a0e1a', width=2),
)
fig_TREE.update_layout(
    paper_bgcolor=P['bg'], font=dict(family='Inter', color=P['text']),
    title_font=dict(size=15, color=P['text'], family='Inter'),
    height=340, margin=dict(l=10, r=10, t=60, b=10),
)
fig_TREE.show()


# ── 2) WoE nedir — açıklama kartı
y_high = (df_filled[TARGET] == 'Yüksek').astype(int)

display(card('BİLGİ', '#22d3ee',
    "WoE (Weight of Evidence) nedir? Her kategorinin log-olasılık oranıdır: "
    "o kategorideki öğrencilerin Yüksek Tükenmişlik grubuna <em>girme</em> olasılığı, "
    "<em>girmeme</em> olasılığına bölünür ve logaritması alınır. "
    "<b>Pozitif WoE</b> = bu kategori ortalamanın üzerinde tükenmişlik riski taşır; "
    "<b>negatif WoE</b> = ortalamanın altında. "
    "Kredi risk modellemesi ve churn prediction'da onlarca yıldır kullanılan yöntemdir."))


# ── 3) WoE & IV hesabı (görselleştirme için manuel)
total_pos = y_high.sum()
total_neg = (1 - y_high).sum()
woe_vis  = {}
iv_data  = {}

for col in WOE_ENC_COLS:
    woe_map = {}
    iv = 0.0
    for cat in sorted(df_filled[col].dropna().unique()):
        mask     = df_filled[col] == cat
        pos      = float(y_high[mask].sum())
        neg      = float((1 - y_high)[mask].sum())
        dist_pos = pos / total_pos
        dist_neg = neg / total_neg
        woe_v    = float(np.log((dist_pos + 1e-9) / (dist_neg + 1e-9)))
        woe_map[cat] = woe_v
        iv      += (dist_pos - dist_neg) * woe_v
    woe_vis[col] = woe_map
    iv_data[col] = iv


# ── 4) WoE çizgi grafiği — kategori → log-odds değeri
_woe_colors = ['#f472b6', '#818cf8']
fig_WOE = make_subplots(rows=1, cols=2,
    subplot_titles=['Okunan Bölüm', 'Birincil Kullanım Amacı'],
    horizontal_spacing=0.12)

for idx, (col, color) in enumerate(zip(WOE_ENC_COLS, _woe_colors), 1):
    woe_map     = woe_vis[col]
    sorted_cats = sorted(woe_map, key=lambda x: woe_map[x])
    cat_labels  = [c.replace('_', ' ') for c in sorted_cats]
    vals        = [woe_map[c] for c in sorted_cats]
    r, g, b     = int(color[1:3], 16), int(color[3:5], 16), int(color[5:7], 16)

    n    = len(cat_labels)
    tpos = (['middle right'] + ['top center'] * (n - 2) + ['middle left']) if n > 1 else ['top center']
    fig_WOE.add_trace(go.Scatter(
        x=cat_labels, y=vals,
        mode='lines+markers+text',
        marker=dict(size=11, color=color, symbol='circle',
                    line=dict(color='rgba(255,255,255,0.25)', width=1.5)),
        line=dict(color=color, width=2.5),
        fill='tozeroy', fillcolor=f'rgba({r},{g},{b},0.10)',
        text=[f'{v:+.3f}' for v in vals],
        textposition=tpos,
        textfont=dict(color=color, size=10, family='Inter'),
        hovertemplate=f'<b>{col}</b><br>%{{x}}<br>WoE: %{{y:+.4f}}<extra></extra>',
        showlegend=False,
    ), row=1, col=idx)

fig_WOE.update_layout(
    title=dict(
        text='<b>WoE Encoding — Kategori → Log-Odds Değeri</b>  '
             '<span style="color:#64748b; font-size:12px">· pozitif = yüksek risk, negatif = düşük risk</span>',
        font=dict(size=16, color=P['text'], family='Inter'), x=0.02, xanchor='left'),
    paper_bgcolor=P['bg'], plot_bgcolor=P['bg'],
    font=dict(family='Inter', color=P['text']),
    height=390, margin=dict(l=60, r=40, t=95, b=120),
)
all_woe_vals = [v for col in WOE_ENC_COLS for v in woe_vis[col].values()]
y_lo = min(all_woe_vals) * 1.55
y_hi = max(all_woe_vals) * 1.55
for j in range(1, 3):
    fig_WOE.update_yaxes(gridcolor=P['grid'], zeroline=True,
                         zerolinecolor='#475569', zerolinewidth=2,
                         tickformat='+.2f', showgrid=True, range=[y_lo, y_hi],
                         tickfont=dict(color=P['subtext'], size=9), row=1, col=j)
    fig_WOE.update_xaxes(showgrid=False, tickangle=-30,
                         tickfont=dict(size=9, color=P['subtext']), row=1, col=j)
fig_WOE.show()


# ── 5) OHE'ye karşı sütun tasarrufu
OHE_WOULD_BE = {
    'Okunan Bölüm': 5, 'Birincil Kullanım Amacı': 5, 'Kurum Politikası': 3,
    'Sınıf Düzeyi': 1, 'Prompt Yazma Becerisi': 1, 'Ücretli Abonelik': 1,
}
SMART = {c: 1 for c in OHE_WOULD_BE}
````

````python
# ── 7) WoE uygulama (category_encoders kütüphanesi)
df_encoded = df_filled.copy()

# Bool → int
df_encoded['Ücretli Abonelik'] = df_encoded['Ücretli Abonelik'].astype(int)

# Ordinal
ord_enc = OrdinalEncoder(categories=ORDINAL_ORDERS,
                         handle_unknown='use_encoded_value', unknown_value=-1)
df_encoded[ORDINAL_COLS] = ord_enc.fit_transform(df_encoded[ORDINAL_COLS])

# WoE Encoding
woe_encoder = WOEEncoder(cols=WOE_ENC_COLS, randomized=False)
woe_encoder.fit(df_filled[WOE_ENC_COLS], y_high)
df_encoded[WOE_ENC_COLS] = woe_encoder.transform(df_filled[WOE_ENC_COLS])

# Sütun sırası: target en sona
cols_ordered = [c for c in df_encoded.columns if c != TARGET] + [TARGET]
df_encoded   = df_encoded[cols_ordered]

assert df_encoded.shape[1] == df_filled.shape[1], "Sütun sayısı değişti!"

# ── 8) Encoded öğrenci matrisi — Plotly heatmap (statik tablo yerine)
ENC_SAMPLE_COLS = WOE_ENC_COLS + ORDINAL_COLS + BOOL_COLS
_sample_enc = df_encoded[ENC_SAMPLE_COLS].head(8).astype(float)

fig_ENC_HM = px.imshow(
    _sample_enc,
    x=[c[:18] + '…' if len(c) > 18 else c for c in ENC_SAMPLE_COLS],
    y=[f'Öğr. {i+1}' for i in range(len(_sample_enc))],
    color_continuous_scale=[
        [0.0, '#7c3aed'], [0.25, '#6d28d9'], [0.5, '#1e293b'],
        [0.75, '#0369a1'], [1.0, '#22d3ee'],
    ],
    aspect='auto',
    text_auto='.3f',
    title='<b>Encoded Öğrenci Matrisi</b> — Satır=öğrenci · Sütun=feature · Renk=kodlanmış değer',
)
fig_ENC_HM.update_traces(textfont=dict(size=9, family='Inter'))
fig_ENC_HM.update_layout(
    paper_bgcolor=P['bg'], font=dict(family='Inter', color=P['text']),
    title_font=dict(size=14, color=P['text'], family='Inter'),
    height=400, margin=dict(l=80, r=20, t=65, b=90),
    coloraxis_colorbar=dict(
        title=dict(text='Değer', font=dict(color=P['subtext'])),
        tickfont=dict(color=P['subtext'], size=9),
    ),
    xaxis=dict(tickfont=dict(size=9, color=P['subtext']), tickangle=-35),
    yaxis=dict(tickfont=dict(size=10, color=P['subtext'])),
)
fig_ENC_HM.show()


# ── 9) Karar kartı
saved = sum(OHE_WOULD_BE.values()) - sum(SMART.values())
display(card('KARAR', '#34d399',
    f"WoE Encoding ile <b>{saved} gereksiz sütundan kurtulduk</b>; veri "
    f"<b>{df_encoded.shape[1]-1} feature</b> ile log-odds biçiminde aynı bilgiyi taşıyor."))

ENCODING_DECISION = {
    'ohe_cols'       : [],
    'ordinal_cols'   : ORDINAL_COLS,
    'ordinal_orders' : ORDINAL_ORDERS,
    'woe_enc_cols'   : WOE_ENC_COLS,
    'bool_cols'      : BOOL_COLS,
    'numeric_cols'   : IMPUTE_NUMERIC,
    'final_features' : df_encoded.shape[1] - 1,
    'ohe_avoided'    : saved,
}
print(f">> Encoding tamam | feature={df_encoded.shape[1]-1} | OHE'ye karşı +{saved} sütun tasarruf")
````

````python
# ════════════════════════════════════════════════════════════════════
# 3.4 — SCALING
# RobustScaler: aykırı değerler korunduğu için medyan/IQR tabanlı seçim.
# ════════════════════════════════════════════════════════════════════
import numpy as np
from sklearn.preprocessing import StandardScaler, RobustScaler, MinMaxScaler

# ── 0) Domain constraint clip
DOMAIN_CLIP = {
    'Haftalık AI Saati'       : (0, None),
    'Geleneksel Çalışma Saati': (0, None),
}
for col, (lo, hi) in DOMAIN_CLIP.items():
    if col in df_encoded.columns:
        df_encoded[col] = df_encoded[col].clip(lower=lo, upper=hi)

# ── 1) Neden scale ediyoruz
display(card('BİLGİ', '#f472b6',
    "Tüm sütunları benzer bir aralığa çekerek dengeliyoruz. Adım 2'de aykırı değerleri "
    "koruma kararı verdik; bu yüzden ortalama tabanlı StandardScaler değil, "
    "<b>medyan ve IQR tabanlı RobustScaler</b> seçiyoruz."))

SCALE_COLS = [c for c in df_encoded.columns
              if c != TARGET
              and c != 'Ücretli Abonelik'
              and df_encoded[c].dtype in ['float64', 'int64']]

# ── 2) 3 Scaler karşılaştırması — dağılım kutu grafiği (Haftalık AI Saati)
demo_col = 'Haftalık AI Saati'
raw_vals = df_encoded[demo_col].values.reshape(-1, 1)

sc_variants = {
    'StandardScaler': (StandardScaler().fit_transform(raw_vals), '#22d3ee', 'rgba(34,211,238,0.18)'),
    'RobustScaler'  : (RobustScaler().fit_transform(raw_vals),   '#34d399', 'rgba(52,211,153,0.18)'),
    'MinMaxScaler'  : (MinMaxScaler().fit_transform(raw_vals),   '#f59e0b', 'rgba(245,158,11,0.18)'),
}

fig_SC = go.Figure()
fig_SC.add_trace(go.Box(
    y=raw_vals.flatten(), name='Orijinal',
    boxpoints='outliers',
    marker=dict(color='#64748b', size=3, opacity=0.4),
    line=dict(color='#64748b', width=1.5),
    fillcolor='rgba(100,116,139,0.15)',
    hovertemplate='Orijinal<br>%{y:.1f}<extra></extra>',
))
for name, (vals_sc, color, fill) in sc_variants.items():
    fig_SC.add_trace(go.Box(
        y=vals_sc.flatten(), name=name,
        boxpoints='outliers',
        marker=dict(color=color, size=3, opacity=0.4),
        line=dict(color=color, width=1.5),
        fillcolor=fill,
        hovertemplate=f'{name}<br>%{{y:.3f}}<extra></extra>',
    ))

fig_SC.update_layout(
    title=dict(
        text='<b>3 Scaler Karşılaştırması — Haftalık AI Saati</b>  '
             '<span style="color:#64748b; font-size:12px">· sağa uzayan aykırıların hangi yöntemde daha az etkili?</span>',
        font=dict(size=16, color=P['text'], family='Inter'), x=0.02, xanchor='left'),
    paper_bgcolor=P['bg'], plot_bgcolor=P['bg'],
    font=dict(family='Inter', color=P['text']),
    xaxis=dict(showgrid=False, tickfont=dict(size=11, color=P['subtext'])),
    yaxis=dict(gridcolor=P['grid'], title='Ölçeklendirilmiş değer'),
    height=420, margin=dict(l=80, r=40, t=90, b=60),
    legend=dict(bgcolor='rgba(15,23,42,0.6)', bordercolor=P['grid'], borderwidth=1,
                orientation='h', y=1.10, x=0.5, xanchor='center'),
)
fig_SC.show()


# ── 3) Scaler istatistik karşılaştırması — Radar Grafiği (örümcek)
stats_rows = []
all_sc_data = [('Orijinal', raw_vals)] +               [(n, v[0]) for n, v in sc_variants.items()]

for name, vals in all_sc_data:
    v = vals.flatten()
    q75, q25 = np.percentile(v, [75, 25])
    stats_rows.append({
        'Yöntem'   : name,
        'Ortalama' : round(float(np.mean(v)), 4),
        'Std'      : round(float(np.std(v)),  4),
        'Min'      : round(float(np.min(v)),  4),
        'Max'      : round(float(np.max(v)),  4),
        'IQR'      : round(float(q75 - q25),  4),
    })

SCALER_COLORS = {
    'Orijinal':       ('rgba(100,116,139,0.15)', '#64748b'),
    'StandardScaler': ('rgba(34,211,238,0.15)',  '#22d3ee'),
    'RobustScaler':   ('rgba(52,211,153,0.15)',  '#34d399'),
    'MinMaxScaler':   ('rgba(245,158,11,0.15)',  '#f59e0b'),
}
_theta = ['Ortalama (abs)', 'Std', 'IQR', 'Maks-Min']

# Normalize 0-1 for radar
_raw_mat = np.array([[abs(r['Ortalama']), r['Std'], r['IQR'], r['Max']-r['Min']]
                      for r in stats_rows])
_col_max  = _raw_mat.max(axis=0)
_norm_mat = _raw_mat / (_col_max + 1e-12)

fig_RADAR = go.Figure()
for _i, row in enumerate(stats_rows):
    vals_n = _norm_mat[_i].tolist()
    closed_r = vals_n + [vals_n[0]]
    closed_t = _theta + [_theta[0]]
    fill_c, line_c = SCALER_COLORS.get(row['Yöntem'], ('rgba(148,163,184,0.12)', '#94a3b8'))
    raw_v = _raw_mat[_i].tolist()
    fig_RADAR.add_trace(go.Scatterpolar(
        r=closed_r, theta=closed_t,
        name=row['Yöntem'],
        fill='toself', fillcolor=fill_c,
        line=dict(color=line_c, width=2.5),
        hovertemplate='<b>' + row['Yöntem'] + '</b><br>'
            + '<br>'.join(f'{_theta[j]}: {raw_v[j]:.4f}' for j in range(4))
            + '<extra></extra>',
    ))
fig_RADAR.update_layout(
    polar=dict(
        bgcolor='rgba(15,23,42,0.85)',
        radialaxis=dict(
            visible=True, range=[0, 1], showticklabels=False,
            gridcolor='rgba(71,85,105,0.4)', gridwidth=1,
        ),
        angularaxis=dict(
            tickfont=dict(size=11, color='#cbd5e1', family='Inter'),
            gridcolor='rgba(71,85,105,0.4)', gridwidth=1,
        ),
    ),
    title=dict(
        text='<b>Scaler Karşılaştırması — Örümcek Grafiği</b>  '
             '<span style="color:#64748b; font-size:12px">· normalize (0-1) · Haftalık AI Saati</span>',
        font=dict(size=15, color=P['text'], family='Inter'), x=0.02, xanchor='left'),
    paper_bgcolor=P['bg'], font=dict(family='Inter', color=P['text']),
    height=440, margin=dict(l=80, r=80, t=80, b=80),
    legend=dict(bgcolor='rgba(15,23,42,0.7)', bordercolor=P['grid'], borderwidth=1,
                font=dict(color=P['text'], size=11)),
)
fig_RADAR.show()

display(card('STRATEJİ', '#34d399',
    "RobustScaler örümcek grafiğinde en küçük <b>Std</b> ve <b>Maks-Min</b> alanını kaplar; "
    "aykırı değerlerin etkisini medyan/IQR ile bastırarak dengeli bir ölçekleme sağlar. "
    "MinMaxScaler aykırı değerleri en çok büzer, StandardScaler ise ortalamayı merkez alır."))
````

      font-size:1.55em; margin:0; letter-spacing:1.2px; text-transform:uppercase;">Feature Engineering</h2>
  </div><div style="color:#94a3b8; font-size:1em; font-weight:400; margin-top:8px; letter-spacing:0.3px;">Mantıksal kombinasyonları sinerji testine sokuyor, sadece yeni bilgi üretenleri pipeline'a ekliyoruz.</div>

````python
# ════════════════════════════════════════════════════════════════════
# 3.5 — ÖZELLİK MÜHENDİSLİĞİ (Feature Engineering)
# Strateji: HER aday için SİNERJİ TESTİ yapıyoruz —
#   MI(kombinasyon) > max( MI(bileşen_1), MI(bileşen_2) ) ?
# Yani: iki ham sütun tek başına zayıfsa, birleştiğinde bunu aşıp
# yeni bilgi üretiyor mu? Eğer evet → ekle; hayır → çöp, mantığı yok.
# ════════════════════════════════════════════════════════════════════
from sklearn.feature_selection import mutual_info_classif
from scipy import stats
from category_encoders import WOEEncoder

# Ortak target hazırlığı
y_cat_fe = df_filled[TARGET]
y_ord_fe = df_filled[TARGET].map({'Düşük':0,'Orta':1,'Yüksek':2})

# (section_header markdown ile değiştirildi)

# ════════════════════════════════════════════════════════════════════
# A) PRE/POST GPA: gerçekten kıymetli mi?
# ════════════════════════════════════════════════════════════════════
display(tbl_header('A) PRE/POST GPA — TAHMİN GÜCÜ TESTİ', '#fbbf24'))

gpa_df = df[['Dönem Öncesi GNO','Dönem Sonrası GNO',TARGET]].dropna()
gpa_df['GNO Ortalaması'] = (gpa_df['Dönem Öncesi GNO'] + gpa_df['Dönem Sonrası GNO']) / 2
gpa_df['GNO Değişimi']  = gpa_df['Dönem Sonrası GNO'] - gpa_df['Dönem Öncesi GNO']
y_gpa = gpa_df[TARGET]

gpa_results = []
for c in ['Dönem Öncesi GNO','Dönem Sonrası GNO','GNO Ortalaması','GNO Değişimi']:
    mi = mutual_info_classif(gpa_df[[c]].values, y_gpa, random_state=42)[0]
    gpa_results.append((c, mi))

ref_col = 'Haftalık AI Saati'
ref_mi  = mutual_info_classif(df_filled[[ref_col]].values, y_cat_fe, random_state=42)[0]

fig_GPA = go.Figure()
fig_GPA.add_trace(go.Bar(
    x=[c for c, _ in gpa_results] + [f'{ref_col} (referans)'],
    y=[m for _, m in gpa_results] + [ref_mi],
    marker=dict(color=['#fbbf24','#fbbf24','#fbbf24','#fbbf24','#22d3ee'], line=dict(width=0)),
    text=[f'{m:.4f}' for _, m in gpa_results] + [f'{ref_mi:.4f}'],
    textposition='outside',
    textfont=dict(color=P['text'], size=11, family='Inter'),
    hovertemplate='%{x}<br>MI=%{y:.5f}<extra></extra>',
    showlegend=False,
))
fig_GPA.update_layout(
    title=dict(text='<b>GPA Tahmin Gücü vs Referans</b>  '
                    '<span style="color:#64748b; font-size:12px">· yüksek MI = tahmin gücü; GPA neredeyse sıfır</span>',
               font=dict(size=16, color=P['text'], family='Inter'), x=0.02, xanchor='left'),
    paper_bgcolor=P['bg'], plot_bgcolor=P['bg'],
    font=dict(family='Inter', color=P['text']),
    xaxis=dict(tickangle=-20, showgrid=False, tickfont=dict(size=10, color=P['subtext'])),
    yaxis=dict(title='Mutual Information', gridcolor=P['grid'], range=[0, ref_mi * 1.25]),
    height=400, margin=dict(l=80, r=40, t=90, b=120),
)
fig_GPA.show()

ratio = ref_mi / max(gpa_results[0][1], 1e-9)
display(card('KARAR', '#fbbf24',
    "Pre/Post GPA hiçbir katkı sağlamıyor — <b>drop ediyoruz</b>."))


# ════════════════════════════════════════════════════════════════════
# B) AKILLI BİRLEŞTİRME — 4 MANTIKSAL ADAY + SİNERJİ TESTİ
# ════════════════════════════════════════════════════════════════════
display(tbl_header('B) MANTIKSAL BİRLEŞTİRME — SİNERJİ TESTİ', '#22d3ee'))

display(card('METODOLOJİ', '#22d3ee',
    "Aday seçiminde tek kural: <b>bileşenler arasında mantıksal motivasyon</b>. "
    "4 aday hipotezi var:<br>"
    "1. <b>Prompt Becerisi × Araç Çeşitliliği</b> → bilinçli kullanıcı profili<br>"
    "2. <b>Ücretli Abonelik × Prompt Becerisi</b> → araca ciddi yatırım yapan kullanıcı<br>"
    "3. <b>Birincil Kullanım Amacı × Okunan Bölüm</b> → bölüm-amaç uyumu (STEM+Coding farklı, Beşeri+Coding farklı)<br>"
    "4. <b>Geleneksel + Haftalık AI Saati</b> → toplam çalışma yükü (ikame mi tamamlayıcı mı?)<br>"
    "Karar kriteri: <code>MI(kombinasyon) > 1.10 × max(MI(bileşen_1), MI(bileşen_2))</code> — "
    "yani kombinasyon, en iyi tek bileşenden <b>en az %10 daha fazla bilgi taşıyor mu?</b>"))

# ── Bileşenlerin tek başına MI'ları (referans için)
mi_single = {}
for c in ['Haftalık AI Saati','Geleneksel Çalışma Saati','Algılanan AI Bağımlılığı',
          'Araç Çeşitliliği','Sınav Kaygı Düzeyi','Beceri Kalıcılık Skoru']:
    mi_single[c] = mutual_info_classif(df_filled[[c]].values.astype(float), y_cat_fe, random_state=42)[0]
for c in ['Prompt Yazma Becerisi','Ücretli Abonelik','Okunan Bölüm','Birincil Kullanım Amacı']:
    mi_single[c] = mutual_info_classif(df_encoded[[c]].values.astype(float), y_cat_fe, random_state=42)[0]

# ── Aday tanımları ve bileşen listesi
prompt_num = df_encoded['Prompt Yazma Becerisi'].astype(float)
paid_num   = df_encoded['Ücretli Abonelik'].astype(float)

# Bölüm × Amaç → string concat → WoE encode (etkileşim WoE'su)
interaction_str = df_filled['Okunan Bölüm'].astype(str) + ' | ' + df_filled['Birincil Kullanım Amacı'].astype(str)
y_high_int = (y_cat_fe == 'Yüksek').astype(int)
_woe_int = WOEEncoder(cols=['BolumAmac'], randomized=False)
_int_df = pd.DataFrame({'BolumAmac': interaction_str})
_woe_int.fit(_int_df, y_high_int)
bolum_amac_woe = _woe_int.transform(_int_df)['BolumAmac'].astype(float).values

candidates = [
    # (ad, vektör, bileşenler-listesi)
    ('Prompt × Araç Çeşitliliği',
        (prompt_num.values + 1) * df_filled['Araç Çeşitliliği'].values,
        ['Prompt Yazma Becerisi', 'Araç Çeşitliliği']),
    ('Abonelik × Prompt Becerisi',
        paid_num.values * (prompt_num.values + 1),
        ['Ücretli Abonelik', 'Prompt Yazma Becerisi']),
    ('Bölüm-Amaç Risk Skoru',
        bolum_amac_woe,
        ['Okunan Bölüm', 'Birincil Kullanım Amacı']),
    ('Toplam Çalışma Yükü (AI + Geleneksel)',
        df_filled['Haftalık AI Saati'].values + df_filled['Geleneksel Çalışma Saati'].values,
        ['Haftalık AI Saati', 'Geleneksel Çalışma Saati']),
]

SYNERGY_RATIO = 1.10  # kombinasyon en iyi bileşenden en az %10 daha bilgi taşımalı

syn_rows = []
for name, vec, comps in candidates:
    mi_combo = mutual_info_classif(np.asarray(vec).reshape(-1,1), y_cat_fe, random_state=42)[0]
    mi_best  = max(mi_single[c] for c in comps)
    sinerji  = mi_combo / max(mi_best, 1e-9)
    karar    = 'EKLE' if sinerji >= SYNERGY_RATIO else 'REDDET'
    syn_rows.append({'Aday': name,
                     'MI (kombo)': mi_combo,
                     'MI (en iyi bileşen)': mi_best,
                     'Sinerji Oranı': sinerji,
                     'Bileşenler': ' + '.join(comps),
                     'Karar': karar})
syn_df = pd.DataFrame(syn_rows).sort_values('Sinerji Oranı', ascending=False).reset_index(drop=True)

# ── Görsel: kombinasyon MI vs en iyi bileşen MI, yan yana
fig_SYN = go.Figure()
fig_SYN.add_trace(go.Bar(
    name='En iyi bileşenin MI\'ı',
    x=syn_df['Aday'], y=syn_df['MI (en iyi bileşen)'],
    marker=dict(color='#475569', line=dict(width=0)),
    text=[f'{v:.4f}' for v in syn_df['MI (en iyi bileşen)']],
    textposition='outside', textfont=dict(color=P['subtext'], size=10, family='Inter'),
    hovertemplate='<b>%{x}</b><br>En iyi bileşen MI=%{y:.5f}<extra></extra>',
))
fig_SYN.add_trace(go.Bar(
    name='Kombinasyonun MI\'ı',
    x=syn_df['Aday'], y=syn_df['MI (kombo)'],
    marker=dict(color=['#34d399' if k=='EKLE' else '#f472b6' for k in syn_df['Karar']], line=dict(width=0)),
    text=[f'{v:.4f}' for v in syn_df['MI (kombo)']],
    textposition='outside', textfont=dict(color=P['text'], size=11, family='Inter'),
    hovertemplate='<b>%{x}</b><br>Kombo MI=%{y:.5f}<extra></extra>',
))
fig_SYN.update_layout(
    barmode='group',
    title=dict(text='<b>Sinerji Testi — Kombinasyon mu Bileşen mi Daha Bilgili?</b>  '
                    '<span style="color:#64748b; font-size:12px">· yeşil = sinerji (EKLE); pembe = redundans (REDDET)</span>',
               font=dict(size=16, color=P['text'], family='Inter'), x=0.02, xanchor='left'),
    paper_bgcolor=P['bg'], plot_bgcolor=P['bg'], font=dict(family='Inter', color=P['text']),
    xaxis=dict(tickangle=-15, showgrid=False, tickfont=dict(size=10, color=P['subtext'])),
    yaxis=dict(title='Mutual Information', gridcolor=P['grid']),
    height=440, margin=dict(l=80, r=40, t=90, b=120),
    legend=dict(bgcolor='rgba(15,23,42,0.6)', bordercolor=P['grid'], borderwidth=1,
                orientation='h', y=1.10, x=0.5, xanchor='center'),
)
fig_SYN.show()

# ── Karar tablosu
display(syn_df.assign(**{
    'MI (kombo)': syn_df['MI (kombo)'].round(4),
    'MI (en iyi bileşen)': syn_df['MI (en iyi bileşen)'].round(4),
    'Sinerji Oranı': syn_df['Sinerji Oranı'].round(2),
}).style
    .set_properties(**{'background-color':'#0f172a','color':'#e2e8f0',
                       'border':'1px solid #1e293b','padding':'8px 14px'})
    .set_table_styles([{'selector':'th','props':[('background-color','#1e293b'),
                                                  ('color','#94a3b8'),('font-weight','700'),
                                                  ('padding','8px 14px')]}])
    .applymap(lambda v: 'color:#34d399;font-weight:700' if v=='EKLE'
              else 'color:#f472b6;font-weight:700' if v=='REDDET' else '',
              subset=['Karar'])
    .hide(axis='index'))

selected_combos = syn_df[syn_df['Karar']=='EKLE']['Aday'].tolist()
rejected_combos = syn_df[syn_df['Karar']=='REDDET']['Aday'].tolist()

if selected_combos:
    display(card('SEÇİLEN (sinerji üretiyor)', '#34d399',
        f"<b>{len(selected_combos)} kombinasyon</b> sinerji testini geçti — "
        f"en iyi bileşeninden en az %10 daha fazla bilgi taşıyor: "
        f"{' · '.join(f'<code>{c}</code>' for c in selected_combos)}"))
if rejected_combos:
    display(card('REDDEDİLEN (redundans)', '#f472b6',
        f"<b>{len(rejected_combos)} kombinasyon</b> sinerji üretmedi — bileşenleri zaten "
        f"yeterli sinyali tek başına taşıyor: "
        f"{' · '.join(f'<code>{c}</code>' for c in rejected_combos)}. "
        "Bunları eklemek modele yeni bilgi katmaz, sadece boyut şişirir."))
````

````python
# ════════════════════════════════════════════════════════════════════
# C) HİPOTEZ TESTLERİ (3 adet — yumuşak dil ile)
# ════════════════════════════════════════════════════════════════════
display(tbl_header('C) HİPOTEZ TESTLERİ — Beklediğimiz İlişkileri Veride Arıyoruz', '#a78bfa'))

# ── H1: Ücretli abonelik → prompt becerisi
display(card('H1', '#a78bfa',
    "<b>Beklenti:</b> Para ödeyip aracı ciddiye alan kullanıcı, prompt becerisini "
    "geliştirmiş olabilir mi? Ücretli abonelik ile prompt becerisi arasında bir "
    "ilişki bekliyoruz."))

ct1 = pd.crosstab(df_filled['Ücretli Abonelik'], df_filled['Prompt Yazma Becerisi'])
chi2_1, p1, _, _ = stats.chi2_contingency(ct1)
ct1_norm = ct1.div(ct1.sum(axis=1), axis=0).reindex(columns=['Başlangıç','Orta','İleri'])

fig_H1 = go.Figure()
skill_colors = {'Başlangıç':'#64748b','Orta':'#a78bfa','İleri':'#22d3ee'}
for skill in ct1_norm.columns:
    fig_H1.add_trace(go.Bar(
        x=['Ücretsiz', 'Ücretli'], y=ct1_norm[skill].values,
        name=skill, marker=dict(color=skill_colors[skill], line=dict(width=0)),
        text=[f'{v*100:.1f}%' for v in ct1_norm[skill].values],
        textposition='inside', textfont=dict(color='#0f172a', size=11, family='Inter'),
        hovertemplate=f'<b>{skill}</b><br>%{{x}}: %{{y:.1%}}<extra></extra>',
    ))
fig_H1.update_layout(barmode='stack',
    title=dict(text='<b>H1: Ücretli Abonelik × Prompt Becerisi Dağılımı</b>',
               font=dict(size=15, color=P['text'], family='Inter'), x=0.02, xanchor='left'),
    paper_bgcolor=P['bg'], plot_bgcolor=P['bg'], font=dict(family='Inter', color=P['text']),
    xaxis=dict(showgrid=False), yaxis=dict(tickformat='.0%', gridcolor=P['grid']),
    height=320, margin=dict(l=80, r=40, t=70, b=50),
    legend=dict(bgcolor='rgba(15,23,42,0.6)', orientation='h', y=1.15, x=0.5, xanchor='center'))
fig_H1.show()

h1_msg = ("<b>Beklenildiği gibi bir ilişki bulunmadı.</b> Ki-kare: χ²={:.2f}, p={:.3f}. "
          "İleri beceri oranı ücretsizde %{:.1f}, ücretlide %{:.1f} — fark anlamsız. "
          "Para ödemek prompt becerisinin habercisi değil; bu iki sinyali kombinasyon olarak "
          "modele eklemek bilgi katmaz.").format(
            chi2_1, p1, ct1_norm.loc['Hayır','İleri']*100 if 'Hayır' in ct1_norm.index else ct1_norm.iloc[0]['İleri']*100,
            ct1_norm.loc['Evet','İleri']*100 if 'Evet' in ct1_norm.index else ct1_norm.iloc[1]['İleri']*100)
display(card('H1 — Sonuç', '#f472b6', h1_msg))

# ── H2: Yüksek AI saati geleneksel çalışmayı ikame ediyor mu?
display(card('H2', '#a78bfa',
    "<b>Beklenti:</b> Öğrenciler haftalık AI kullanım saatini artırdıkça geleneksel "
    "çalışma saatlerini düşürüyor olabilir mi (ikame ilişkisi)? Negatif Spearman "
    "korelasyonu bekliyoruz."))

rho_h2, p_h2 = stats.spearmanr(df_filled['Haftalık AI Saati'], df_filled['Geleneksel Çalışma Saati'])

fig_H2 = go.Figure()
fig_H2.add_trace(go.Histogram2dContour(
    x=df_filled['Haftalık AI Saati'], y=df_filled['Geleneksel Çalışma Saati'],
    colorscale=[[0,'#0f172a'],[0.5,'#7c3aed'],[1.0,'#22d3ee']],
    showscale=False, contours=dict(showlines=False),
    hovertemplate='AI Saati=%{x:.1f}<br>Geleneksel=%{y:.1f}<extra></extra>',
))
fig_H2.update_layout(
    title=dict(text=f'<b>H2: AI Saati × Geleneksel Çalışma — Yoğunluk Haritası</b>  '
                    f'<span style="color:#64748b;font-size:12px">· Spearman ρ={rho_h2:+.3f}, p={p_h2:.3f}</span>',
               font=dict(size=15, color=P['text'], family='Inter'), x=0.02, xanchor='left'),
    paper_bgcolor=P['bg'], plot_bgcolor=P['bg'], font=dict(family='Inter', color=P['text']),
    xaxis=dict(title='Haftalık AI Saati', gridcolor=P['grid']),
    yaxis=dict(title='Geleneksel Çalışma Saati', gridcolor=P['grid']),
    height=400, margin=dict(l=80, r=40, t=70, b=50))
fig_H2.show()

h2_iliski = abs(rho_h2) > 0.05 and p_h2 < 0.05
h2_sign = "negatif" if rho_h2 < 0 else "pozitif"
if h2_iliski and rho_h2 < 0:
    h2_msg = (f"<b>Beklenildiği gibi negatif yönlü bir ilişki bulundu.</b> Spearman ρ={rho_h2:+.3f}, p={p_h2:.3f}. "
              "AI kullanımı arttıkça geleneksel çalışma azalıyor — ikame ilişkisi mevcut. "
              "Toplam Çalışma Yükü kombinasyonu bu nedenle hâlâ anlamlı olabilir.")
    h2_color = '#34d399'
else:
    h2_msg = (f"<b>Bir ilişki beklerken bulunamadı.</b> Spearman ρ={rho_h2:+.3f} ({h2_sign}, neredeyse sıfır), p={p_h2:.3f}. "
              "Öğrenciler AI'yı geleneksel çalışmayla <b>paralel</b> kullanıyor — ikame değil ek yük. "
              "Bu, 'Toplam Çalışma Yükü' kombinasyonunun anlamlı olabileceğini destekler: "
              "iki bağımsız zaman havuzunun toplamı, yorgunluğun gerçek göstergesi olur.")
    h2_color = '#fbbf24'
display(card('H2 — Sonuç', h2_color, h2_msg))

# ── H3: STEM öğrencileri Coding/Debugging'i daha mı yoğun kullanıyor?
display(card('H3', '#a78bfa',
    "<b>Beklenti:</b> Fen-Mühendislik (STEM) öğrencileri Birincil Kullanım Amacı olarak "
    "kodlama/hata ayıklamayı, Beşeri öğrenciler ise metin yazımı/özetlemeyi tercih "
    "ediyor olabilir mi? Bölüm × Amaç etkileşim sinyalinin temelini sınıyoruz."))

ct3 = pd.crosstab(df_filled['Okunan Bölüm'], df_filled['Birincil Kullanım Amacı'])
chi2_3, p3, _, _ = stats.chi2_contingency(ct3)
ct3_norm = ct3.div(ct3.sum(axis=1), axis=0)

fig_H3 = px.imshow(
    ct3_norm.values,
    x=ct3_norm.columns, y=ct3_norm.index,
    color_continuous_scale=[[0,'#0f172a'],[0.5,'#7c3aed'],[1.0,'#22d3ee']],
    aspect='auto', text_auto='.1%',
)
fig_H3.update_layout(
    title=dict(text=f'<b>H3: Bölüm × Birincil Kullanım Amacı — Satır Yüzdeleri</b>  '
                    f'<span style="color:#64748b;font-size:12px">· χ²={chi2_3:.1f}, p={p3:.4f}</span>',
               font=dict(size=15, color=P['text'], family='Inter'), x=0.02, xanchor='left'),
    paper_bgcolor=P['bg'], plot_bgcolor=P['bg'], font=dict(family='Inter', color=P['text']),
    xaxis=dict(tickangle=-25, side='bottom'), height=420,
    margin=dict(l=160, r=80, t=70, b=140),
    coloraxis_colorbar=dict(title='Oran', tickformat='.0%'))
fig_H3.show()

if p3 < 0.05:
    h3_msg = (f"<b>Beklenildiği gibi anlamlı bir ilişki bulundu.</b> χ²={chi2_3:.1f}, p={p3:.4f}. "
              "Bölüm ile kullanım amacı bağımsız değil — STEM ve Beşeri öğrencilerin AI kullanım profili farklılaşıyor. "
              "Bu, B bölümündeki <b>Bölüm × Amaç</b> etkileşim feature'ının somut temelidir.")
    h3_color = '#34d399'
else:
    h3_msg = (f"<b>Bir ilişki beklerken bulunamadı.</b> χ²={chi2_3:.1f}, p={p3:.4f}. "
              "Tüm bölümler benzer kullanım amaçlarını sergiliyor — etkileşim feature'ı zayıf çıkacaktır.")
    h3_color = '#fbbf24'
display(card('H3 — Sonuç', h3_color, h3_msg))
````

````python
# ════════════════════════════════════════════════════════════════════
# D) AKILLI SİLME — Tek başına çok zayıf MI taşıyan sütunlar
# ════════════════════════════════════════════════════════════════════
display(tbl_header('D) AKILLI SİLME — Tek Başına Anlamlı Sinyal Üretmeyen Sütunlar', '#f472b6'))

drop_cols = []

# Aday 1: Öğrenci No (kimlik) — varsa
if 'Öğrenci No' in df_scaled.columns:
    sid_mi = mutual_info_classif(df_scaled[['Öğrenci No']].values, y_cat_fe, random_state=42)[0]
    drop_cols.append(('Öğrenci No', sid_mi,
                      'Benzersiz kimlik — tahmin gücü sıfır, modelin sahte örüntü öğrenmesine yol açar.'))

# Aday 2: Tek başına çok zayıf MI taşıyan sütunlar
WEAK_MI_THR = 0.005
weak_candidates = ['Kurum Politikası', 'Ücretli Abonelik', 'Araç Çeşitliliği']
for col in weak_candidates:
    if col in df_scaled.columns:
        mi_val = mutual_info_classif(df_scaled[[col]].values, y_cat_fe, random_state=42)[0]
        if mi_val < WEAK_MI_THR:
            drop_cols.append((col, mi_val,
                f'Tek başına tahmin gücü çok zayıf (MI={mi_val:.4f} < {WEAK_MI_THR}); '
                'kombinasyon adaylarında da sinerji üretmedi.'))

if drop_cols:
    rows = [{'Sütun': c, 'MI': f'{mi:.6f}', 'Neden': reason} for c, mi, reason in drop_cols]
    drop_df = pd.DataFrame(rows)
    display(drop_df.style
        .set_properties(**{'background-color':'#0f172a','color':'#e2e8f0',
                           'border':'1px solid #1e293b','padding':'11px 16px',
                           'font-size':'0.96em'})
        .set_properties(subset=['Neden'], **{'text-align':'left'})
        .set_table_styles([{'selector':'th','props':[('background-color','#1e293b'),
                                                      ('color','#f472b6'),('font-weight','800'),
                                                      ('padding','11px 16px'),
                                                      ('letter-spacing','1px'),
                                                      ('font-size','0.78em'),
                                                      ('text-transform','uppercase')]}])
        .hide(axis='index'))
    display(card('SİLİNDİ', '#f472b6',
        f"<b>{len(drop_cols)} sütun silindi.</b> Eşik: MI < {WEAK_MI_THR}. "
        "Bu sütunlar tek başına anlamlı sinyal taşımıyor ve kombinasyon testlerinde de sinerji üretmediler."))
else:
    display(card('GÖZLEM', '#34d399',
        "Tüm sütunlar en azından eşik MI'sını aştı — silinecek zayıf sütun yok."))


# ════════════════════════════════════════════════════════════════════
# E) UYGULAMA — Seçilen kombinasyonları ekle; birleştirilen orijinalleri kaldır
# ════════════════════════════════════════════════════════════════════
display(tbl_header('E) UYGULAMA — Sinerji Testini Geçenler Pipeline\'a Ekleniyor', '#34d399'))

df_features = df_scaled.drop(columns=[c for c, _, _ in drop_cols], errors='ignore').copy()

cand_map = {name: vec   for name, vec, _    in candidates}
comp_map = {name: comps for name, _, comps in candidates}
selected_features = {name: np.asarray(cand_map[name], dtype=float) for name in selected_combos}

# Kombinasyon eklendiğinde bileşenlerini de pipeline'dan çıkar — bilgi zaten kombinasyonda
absorbed_components = set()
for name in selected_combos:
    for comp in comp_map[name]:
        absorbed_components.add(comp)

from sklearn.preprocessing import RobustScaler
if selected_features:
    new_scaler = RobustScaler()
    new_cols   = list(selected_features.keys())
    new_df = pd.DataFrame(selected_features, index=df_filled.index)
    new_df[new_cols] = new_scaler.fit_transform(new_df[new_cols])

    to_drop_components = [c for c in absorbed_components if c in df_features.columns]
    if to_drop_components:
        df_features = df_features.drop(columns=to_drop_components)

    target_col = df_features[TARGET]
    df_features = df_features.drop(columns=[TARGET])
    for c in new_cols:
        df_features[c] = new_df[c].values
    df_features[TARGET] = target_col.values

    absorbed_msg = (f" Bileşen sütunları da kaldırıldı ({' · '.join(f'<code>{c}</code>' for c in to_drop_components)}) "
                    f"— bilgi artık kombinasyonda." if to_drop_components else "")
    display(card('UYGULANDI', '#34d399',
        f"<b>+ {len(new_cols)} yeni feature</b> RobustScaler ile ölçeklendirildi ve eklendi: "
        + " · ".join(f'<code>{c}</code>' for c in new_cols)
        + f".{absorbed_msg} Toplam feature: <b>{df_features.shape[1]-1}</b> (önceki: {df_scaled.shape[1]-1})."))
else:
    new_cols = []
    to_drop_components = []
    display(card('UYGULANDI', '#fbbf24',
        "Sinerji testini geçen aday olmadı — yeni feature eklenmiyor. Ham feature seti yeterli sinyali taşıyor."))
````

````python
# ════════════════════════════════════════════════════════════════════
# F) NİHAİ KORELASYON MATRİSİ — sıfıra yakınlar soluk
# ════════════════════════════════════════════════════════════════════
display(tbl_header('F) NİHAİ KORELASYON MATRİSİ — Redundans Denetimi', '#f472b6'))

feature_cols = [c for c in df_features.columns if c != TARGET]
corr = df_features[feature_cols].corr().round(2)

# Sıfıra yakın değerleri soluk göstermek için custom text rendering
abs_corr = corr.abs()
text_matrix = corr.values.copy().astype(object)
text_colors = np.where(abs_corr.values < 0.10, 'rgba(226,232,240,0.25)', 'rgba(226,232,240,1.0)')

fig_C = go.Figure(data=go.Heatmap(
    z=corr.values, x=corr.columns, y=corr.index,
    colorscale=[
        [0.0,  '#f472b6'], [0.25, '#7c3aed'],
        [0.5,  '#0f172a'],
        [0.75, '#0891b2'], [1.0,  '#22d3ee'],
    ],
    zmid=0, zmin=-1, zmax=1,
    text=[[f'{v:.2f}' if abs(v) >= 0.10 else f'<span style="opacity:0.25">{v:.2f}</span>'
           for v in row] for row in corr.values],
    texttemplate='%{text}',
    textfont=dict(size=10, family='Inter', color='#e2e8f0'),
    hovertemplate='%{y} × %{x}<br>r = %{z:.3f}<extra></extra>',
    colorbar=dict(title=dict(text='r', font=dict(color=P['text'])),
                  tickfont=dict(color=P['subtext'], size=10), len=0.7,
                  thickness=14, tickformat='+.1f'),
    xgap=2, ygap=2,
))
fig_C.update_layout(
    title=dict(text='<b>Final Feature Korelasyon Matrisi</b>  '
                    '<span style="color:#64748b; font-size:12px">· |r|<0.10 soluk gösterildi (anlamsız gürültü) · |r|>0.85 = redundans</span>',
               font=dict(size=16, color=P['text'], family='Inter'), x=0.02, xanchor='left'),
    paper_bgcolor=P['bg'], plot_bgcolor=P['bg'], font=dict(family='Inter', color=P['text']),
    xaxis=dict(tickangle=-40, tickfont=dict(size=11, color=P['subtext']), side='bottom'),
    yaxis=dict(tickfont=dict(size=11, color=P['subtext']), autorange='reversed'),
    height=750, margin=dict(l=210, r=80, t=100, b=210),
)
fig_C.show()

high_corr_pairs = []
for i in range(len(feature_cols)):
    for j in range(i+1, len(feature_cols)):
        r = corr.iloc[i, j]
        if abs(r) > 0.85:
            high_corr_pairs.append((feature_cols[i], feature_cols[j], r))

if high_corr_pairs:
    pair_str = ' · '.join(f'<code>{a}</code>↔<code>{b}</code> (r={r:+.2f})' for a, b, r in high_corr_pairs)
    display(card('UYARI', '#f472b6',
        f"|r| > 0.85 olan {len(high_corr_pairs)} çift bulundu: {pair_str}. "
        "Tree-based modeller için sorun değil; LogisticRegression için L2 regularization yönetir."))
else:
    display(card('TEMİZ', '#34d399',
        "|r| > 0.85 olan feature çifti yok — feature seti redundans açısından temiz."))


# ════════════════════════════════════════════════════════════════════
# FINAL ÖZET
# ════════════════════════════════════════════════════════════════════
FE_DECISION = {
    'gpa_dropped'     : ['Dönem Öncesi GNO', 'Dönem Sonrası GNO'],
    'gpa_reason'      : f'MI ≤ 0.005 (referansın 1/{ratio:.0f}\'i); birleştirmeler de sinyal üretmedi',
    'dropped_in_fe'   : [c for c, _, _ in drop_cols],
    'synergy_ratio'   : SYNERGY_RATIO,
    'added_features'  : new_cols,
    'rejected_combos' : rejected_combos,
    'hypotheses'      : {
        'H1_paid_skill'  : {'chi2': float(chi2_1), 'p': float(p1)},
        'H2_ai_traditional': {'rho': float(rho_h2), 'p': float(p_h2)},
        'H3_dept_usecase': {'chi2': float(chi2_3), 'p': float(p3)},
    },
    'final_shape'     : df_features.shape,
    'high_corr_pairs' : [(a, b, float(r)) for a, b, r in high_corr_pairs],
}

print(f">> Feature Engineering tamam | shape: {df_features.shape} | +{len(new_cols)} yeni | -{len(drop_cols)} zayıf drop | -{len(to_drop_components)} bileşen absorbe | -2 GPA (cell 26'da)")
display(card('SONUÇ', '#34d399',
    f"Pre/Post GPA verile silindi; Öğrenci No silindi; sinerji testini geçen "
    f"<b>{len(new_cols)} kombinasyon</b> eklendi"
    + (f" ({' · '.join(new_cols)})" if new_cols else "")
    + f". Final feature seti: <b>{df_features.shape[1]-1} sütun</b>. "
    "Bir sonraki adım: Özellik Seçimi."))
````

````python
# ════════════════════════════════════════════════════════════════════
# 3.6 — ÖZELLİK SEÇİMİ (Feature Selection)
# Strateji: 2 bağımsız perspektif —
#   (1) MI       — bilgi-teorik (tek değişken sinyali, model bağımsız)
#   (2) Importance — RandomForest (etkileşimleri yakalar, model bağımlı)
# Karar: BİR özellik HEM MI eşiğini HEM Importance eşiğini aşamıyorsa silinir.
# Varyans bilgi amaçlı raporlanır (RobustScaler sonrası eleme yapmaz).
# RFE kaldırıldı: doğrusal model bağımlı, MI ile çakışıyor.
# ════════════════════════════════════════════════════════════════════
from sklearn.feature_selection import mutual_info_classif
from sklearn.ensemble import RandomForestClassifier

X = df_features.drop(columns=[TARGET])
y = df_features[TARGET]
feature_cols = list(X.columns)
ENGINEERED = set(FE_DECISION['added_features'])

display(card('BİLGİ', '#a78bfa',
    "Özellik seçimi nedir? Mühendislikte feature türettik; şimdi <b>2 bağımsız metrikle</b> "
    "her feature'ı puanlıyoruz:<br>"
    "• <b>MI</b> — Mutual Information: özelliğin target hakkında taşıdığı bilgi (model bağımsız)<br>"
    "• <b>Importance</b> — RandomForest önem skoru (etkileşimleri yakalar, model bağımlı)<br>"
    "Karar kuralı: <code>MI < MI_eşik AND Importance < IMP_eşik</code> → SİL, aksi halde TUT. "
    "Bir feature'ı silmek için <b>her iki yöntemin de</b> ona değer vermiyor olması şart. "
    "Varyans bilgi amaçlı raporlanır (RobustScaler sonrası elemeye yetkin değil)."))

# ── Hesapla
mi_scores = dict(zip(feature_cols, mutual_info_classif(X.values, y, random_state=42)))
variances = X.var().to_dict()
rf = RandomForestClassifier(n_estimators=200, max_depth=10, random_state=42, n_jobs=-1).fit(X, y)
fi_scores = dict(zip(feature_cols, rf.feature_importances_))

# Veri-bağımlı eşikler: her iki metriğin medyanın %50'si
MI_THR  = max(0.01, np.median(list(mi_scores.values())) * 0.50)
IMP_THR = np.median(list(fi_scores.values())) * 0.50
VAR_THR = 0.01

# ════════════════════════════════════════════════════════════════════
# 1) MUTUAL INFORMATION — Hedefe Katkı Grafiği
# ════════════════════════════════════════════════════════════════════
display(tbl_header('1) MUTUAL INFORMATION — Tükenmişliğe Katkı', '#22d3ee'))

mi_sorted = sorted(mi_scores.items(), key=lambda x: x[1])
def mi_color(c, v):
    if c in ENGINEERED: return '#22d3ee'
    return '#34d399' if v >= MI_THR else '#475569'

fig_MI = go.Figure(go.Bar(
    y=[c for c, _ in mi_sorted], x=[v for _, v in mi_sorted], orientation='h',
    marker=dict(color=[mi_color(c, v) for c, v in mi_sorted]),
    text=[f'{v:.4f}' for _, v in mi_sorted], textposition='outside',
    textfont=dict(color=P['text'], size=10, family='Inter'),
    hovertemplate='%{y}<br>MI=%{x:.5f}<extra></extra>'))
fig_MI.add_vline(x=MI_THR, line=dict(color='#f472b6', width=1.5, dash='dash'),
                 annotation_text=f' MI eşiği = {MI_THR:.3f}', annotation_position='top right',
                 annotation=dict(font=dict(color='#f472b6', size=10)))
fig_MI.update_layout(
    title=dict(text='<b>MI — Tükenmişliğe Katkı (yüksek = daha bilgili)</b>  '
                    '<span style="color:#64748b;font-size:12px">· cyan = türetilmiş feature · gri = eşik altı</span>',
               font=dict(size=15, color=P['text'], family='Inter'), x=0.02, xanchor='left'),
    paper_bgcolor=P['bg'], plot_bgcolor=P['bg'], font=dict(family='Inter', color=P['text']),
    xaxis=dict(title='Mutual Information', gridcolor=P['grid'], range=[0, max(mi_scores.values())*1.18]),
    yaxis=dict(showgrid=False, tickfont=dict(size=10, color=P['subtext'])),
    height=520, margin=dict(l=240, r=70, t=70, b=50))
fig_MI.show()


# ════════════════════════════════════════════════════════════════════
# 2) VARYANS — Yeterince Değişken mi? (bilgi amaçlı)
# ════════════════════════════════════════════════════════════════════
display(tbl_header('2) VARYANS DENETİMİ — Bilgi Amaçlı', '#fbbf24'))

var_sorted = sorted(variances.items(), key=lambda x: x[1])
fig_VAR = go.Figure(go.Bar(
    y=[c for c, _ in var_sorted], x=[v for _, v in var_sorted], orientation='h',
    marker=dict(color=['#475569' if v < VAR_THR else '#fbbf24' for _, v in var_sorted]),
    text=[f'{v:.2f}' for _, v in var_sorted], textposition='outside',
    textfont=dict(color=P['text'], size=10, family='Inter'),
    hovertemplate='%{y}<br>Varyans=%{x:.3f}<extra></extra>'))
fig_VAR.update_layout(
    title=dict(text='<b>Varyans — Yeterince Değişken mi?</b>  '
                    '<span style="color:#64748b;font-size:12px">· RobustScaler sonrası hepsi eşiği geçer — bu yöntem kimseyi elemez</span>',
               font=dict(size=15, color=P['text'], family='Inter'), x=0.02, xanchor='left'),
    paper_bgcolor=P['bg'], plot_bgcolor=P['bg'], font=dict(family='Inter', color=P['text']),
    xaxis=dict(title='Varyans (log)', gridcolor=P['grid'], type='log'),
    yaxis=dict(showgrid=False, tickfont=dict(size=10, color=P['subtext'])),
    height=520, margin=dict(l=240, r=70, t=70, b=50))
fig_VAR.show()

n_low_var = sum(1 for v in variances.values() if v < VAR_THR)
display(card('GÖZLEM', '#fbbf24',
    f"Varyans eşiğinin (<{VAR_THR}) altında <b>{n_low_var} feature</b> bulundu. "
    "RobustScaler tüm sayısal sütunları benzer ölçekte tutuyor; varyansla eleme yapmıyoruz "
    "ama küçük varyans bilgi düşüklüğüne işaret edebilir — MI ve Importance bunu zaten yakalıyor."))


# ════════════════════════════════════════════════════════════════════
# 3) RANDOM FOREST IMPORTANCE
# ════════════════════════════════════════════════════════════════════
display(tbl_header('3) RANDOM FOREST IMPORTANCE — Etkileşim-Duyarlı Önem', '#f472b6'))

fi_sorted = sorted(fi_scores.items(), key=lambda x: x[1])
fig_FI = go.Figure(go.Bar(
    y=[c for c, _ in fi_sorted], x=[v for _, v in fi_sorted], orientation='h',
    marker=dict(color=['#22d3ee' if c in ENGINEERED
                       else ('#f472b6' if v >= IMP_THR else '#475569')
                       for c, v in fi_sorted]),
    text=[f'{v:.3f}' for _, v in fi_sorted], textposition='outside',
    textfont=dict(color=P['text'], size=10, family='Inter'),
    hovertemplate='%{y}<br>Önem=%{x:.4f}<extra></extra>'))
fig_FI.add_vline(x=IMP_THR, line=dict(color='#34d399', width=1.5, dash='dash'),
                 annotation_text=f' Importance eşiği = {IMP_THR:.3f}', annotation_position='top right',
                 annotation=dict(font=dict(color='#34d399', size=10)))
fig_FI.update_layout(
    title=dict(text='<b>RandomForest Önem Skoru</b>  '
                    '<span style="color:#64748b;font-size:12px">· cyan = türetilmiş · pembe = eşik üstü · gri = eşik altı</span>',
               font=dict(size=15, color=P['text'], family='Inter'), x=0.02, xanchor='left'),
    paper_bgcolor=P['bg'], plot_bgcolor=P['bg'], font=dict(family='Inter', color=P['text']),
    xaxis=dict(title='Önem Skoru', gridcolor=P['grid'], range=[0, max(fi_scores.values())*1.18]),
    yaxis=dict(showgrid=False, tickfont=dict(size=10, color=P['subtext'])),
    height=520, margin=dict(l=240, r=90, t=70, b=50))
fig_FI.show()


# ════════════════════════════════════════════════════════════════════
# BİRLEŞİK KARAR TABLOSU — MI + Importance Eşik-Tabanlı
# ════════════════════════════════════════════════════════════════════
display(tbl_header('NİHAİ KARAR — MI + Importance Eşik-Tabanlı', '#34d399'))

rows = []
for c in sorted(mi_scores, key=lambda x: -mi_scores[x]):
    mi_ok  = mi_scores[c] >= MI_THR
    imp_ok = fi_scores[c] >= IMP_THR
    karar  = 'TUT' if (mi_ok or imp_ok) else 'SİL'
    rows.append({
        'Özellik'   : c,
        'Kaynak'    : 'Türetilmiş' if c in ENGINEERED else 'Ham',
        'MI'        : round(mi_scores[c], 4),
        'MI ≥ eşik' : '✓' if mi_ok else '·',
        'Importance': round(fi_scores[c], 4),
        'Imp ≥ eşik': '✓' if imp_ok else '·',
        'KARAR'     : karar,
    })
vote_df = pd.DataFrame(rows)

def _karar_c(v):
    return ('color:#34d399;font-weight:700;background-color:rgba(52,211,153,0.10)' if v == 'TUT'
            else 'color:#f472b6;font-weight:700;background-color:rgba(244,114,182,0.10)')
def _kaynak_c(v):
    return 'color:#22d3ee;font-weight:600' if v == 'Türetilmiş' else 'color:#94a3b8'
def _chk_c(v): return 'color:#34d399;font-weight:700' if v == '✓' else 'color:#475569'

display(vote_df.style
    .set_properties(**{'background-color':'#0f172a','color':'#e2e8f0',
                       'border':'1px solid #1e293b','padding':'7px 14px'})
    .set_table_styles([{'selector':'th','props':[('background-color','#1e293b'),
                                                  ('color','#94a3b8'),('font-weight','700'),
                                                  ('padding','7px 14px')]}])
    .applymap(_kaynak_c, subset=['Kaynak'])
    .applymap(_chk_c, subset=['MI ≥ eşik','Imp ≥ eşik'])
    .applymap(_karar_c, subset=['KARAR'])
    .hide(axis='index'))

final_keep = [r['Özellik'] for r in rows if r['KARAR'] == 'TUT']
final_drop = [r['Özellik'] for r in rows if r['KARAR'] == 'SİL']
df_selected = df_features[final_keep + [TARGET]]
n_eng_kept = len([c for c in final_keep if c in ENGINEERED])

display(card('GÖZLEM', '#34d399',
    f"<b>{len(final_keep)} özellik tutuldu, {len(final_drop)} silindi.</b> "
    + (f"Silinen: <b>{', '.join(final_drop)}</b> — hem MI hem Importance eşiğinin altında, "
       "yani iki bağımsız yöntem de tükenmişlik tahmininde bu sütunlara değer vermiyor. "
       if final_drop else "Hiçbir feature elenmedi — her biri en az bir yöntemin eşiğini geçti. ")
    + (f"Türettiğimiz <b>{n_eng_kept}/{len(ENGINEERED)}</b> kombinasyon final sette."
       if ENGINEERED else "")
    + " Karar kuralı tek bir net cümle: <em>'iki yöntem de değer vermiyorsa sil.'</em>"))

display(tbl_header('SEÇİLEN ÖZELLİKLER — Modele Girecek Set', '#22d3ee'))
sel_df = pd.DataFrame({
    '#'       : range(1, len(final_keep)+1),
    'Özellik' : final_keep,
    'Kaynak'  : ['Türetilmiş' if c in ENGINEERED else 'Ham' for c in final_keep],
    'MI'      : [round(mi_scores[c], 4) for c in final_keep],
    'Önem'    : [round(fi_scores[c], 4) for c in final_keep],
})
display(sel_df.style
    .set_properties(**{'background-color':'#0f172a','color':'#e2e8f0',
                       'border':'1px solid #1e293b','padding':'7px 16px'})
    .set_table_styles([{'selector':'th','props':[('background-color','#1e293b'),
                                                  ('color','#94a3b8'),('font-weight','700'),
                                                  ('padding','7px 16px')]}])
    .applymap(_kaynak_c, subset=['Kaynak'])
    .hide(axis='index'))

FS_DECISION = {
    'method'          : 'esik_tabanli_mi_AND_importance',
    'mi_thr'          : float(MI_THR),
    'imp_thr'         : float(IMP_THR),
    'rule'            : 'SİL: MI < MI_thr AND Importance < IMP_thr',
    'mi_scores'       : {c: round(float(v), 6) for c, v in mi_scores.items()},
    'fi_scores'       : {c: round(float(v), 6) for c, v in fi_scores.items()},
    'final_features'  : final_keep,
    'dropped_features': final_drop,
    'final_shape'     : df_selected.shape,
}
print(f">> Feature Selection tamam | {len(feature_cols)} → {len(final_keep)} özellik | "
      f"df_selected: {df_selected.shape} | türetilmiş tutulan: {n_eng_kept}/{len(ENGINEERED)}")
````
